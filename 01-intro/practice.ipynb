```python
V=np.array([[1, 0], [0, 1], [2, 3]])
U = np.array([[1, 2, 3], [4, 5, 6]])
def matmat(U, V):
    result = []
    for i in range(V.shape[1]):     # V.shape is (3, 2); [1] picks the 2 = number of columns
        column = V[:, i]            # ":" = all rows, "i" = column i → [1, 0, 2], then [0, 1, 3]
        result.append(matvec(U, column))   # U times this column → one column of the answer
    return np.array(result).T       # results were stacked as rows; .T flips them into columns
matmat(U,V)
```

```python
print(X.shape)            # 10,2
print(X.T.shape)          # 2,10
print((X.T @ X).shape)    # 2,10 @ 10,2 (inner same, outer stays and inner discarded)
print((X @ X.T).shape)    # ? 10,2 @ 2,10 ← the other order. What happens?
```

```python
w=(np.linalg.inv(X.T@X)@X.T@Y)
print(w)
print (Y.shape)
```

```python
y_pred=X@w  #10,2 * 2,1
print(y_pred)
```

```python
print(y_pred.round(2))
print(Y.round(2))


print(np.mean(np.abs(y_pred-Y)))
```

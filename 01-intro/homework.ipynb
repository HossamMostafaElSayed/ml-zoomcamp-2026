```python
data_filtered=data_asia[["vehicle_weight","model_year"]].head(7)
```

```python

X=data_filtered[["vehicle_weight","model_year"]].values
print(X)
```

```python
XTX=(X.T@X)
#2,7 * 7,2
```

```python
invXTX=np.linalg.inv(XTX)
```

```python
y=[1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
print(y)
```

```python
print(XTX)
```

```python
w=invXTX@X.T@y
```
